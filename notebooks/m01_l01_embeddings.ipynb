## Embed

```
from agentlab.embed import AXES, cosine, embed

a = embed("I want my money back")
b = embed("How do refunds work?")
c = embed("When will my parcel arrive?")

print("axes:", ", ".join(AXES[:6]), "...")
print("money back vs refunds:", round(cosine(a, b), 2))
print("money back vs parcel: ", round(cosine(a, c), 2))
```
