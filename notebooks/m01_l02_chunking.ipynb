## Chunk

```
from collections import Counter

from agentlab.retrieve import load_chunks

chunks = load_chunks()
print("chunks:", len(chunks))
for doc, n in sorted(Counter(c.doc for c in chunks).items()):
    print(f"  {doc:<11} {n} chunks")
first = chunks[0]
print("example:", first.cite, "-", first.text[:36], "...")
```
