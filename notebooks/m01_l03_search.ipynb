## Search

```
from agentlab.retrieve import VectorIndex, load_chunks

index = VectorIndex(load_chunks())    # embed every chunk once
query = "When do I get my money back?"
for score, chunk in index.search(query, k=3):
    print(f"{score:.2f}  {chunk.cite:<11} {chunk.text[:40]}...")
```
