## Rag

```
import textwrap

from agentlab.rag import SEARCH_DOCS, RagTeachingModel
from shopassist.agent import run_agent

SYSTEM = ("Answer only from the passages search_docs returns. Cite the "
          "source in brackets. If nothing relevant comes back, say so.")
for q in ["How long does a refund take?", "Do you sell zebras?"]:
    answer, _ = run_agent(RagTeachingModel(), [SEARCH_DOCS], SYSTEM, q)
    print("Q:", q)
    print(textwrap.fill("A: " + answer, 68, subsequent_indent="   "))
```
