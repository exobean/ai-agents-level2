## Quality

```
from agentlab.rag import INDEX
from agentlab.ragevals import EVALSET, hit_at_k

print("labelled questions:", len(EVALSET))
for k in (1, 2, 3):
    rate, misses = hit_at_k(INDEX, k)
    print(f"hit@{k}: {rate:.0%}  misses: {len(misses)}")
print("still missed:", hit_at_k(INDEX, 3)[1])
```
