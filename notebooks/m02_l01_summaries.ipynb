## Compact

```
import textwrap

from agentlab.memory import compact, tokens
from shopassist.messages import Message

msgs = []
for i in range(10):
    msgs += [Message("user", f"Any news on order #104{i % 4}?"),
             Message("assistant", "It leaves the warehouse tomorrow.")]
small = compact(msgs, keep_last=4)
print("messages:", len(msgs), "->", len(small))
print("tokens:  ", tokens(msgs), "->", tokens(small))
print(textwrap.fill(small[0].text, 68))
```
