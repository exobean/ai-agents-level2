## Memory

```
from agentlab.memory import MemoryStore

store = MemoryStore()
for fact in ["Prefers UPI payment", "Is a Plus member",
             "Card number 4111 1111 1111 1111"]:
    r = store.remember("cust-7", fact)
    verdict = "saved" if r["saved"] else "refused: " + r["reason"]
    print(f"{fact:<34}", verdict)

# weeks later, a new conversation:
print("recall:", store.recall("cust-7", "how do they like to pay?"))
print("forget:", store.forget("cust-7"), "facts deleted")
```
