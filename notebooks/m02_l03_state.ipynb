## Flow

```
from agentlab.workflow import InvalidStep, ReturnFlow

flow = ReturnFlow()
print("state:", flow.state, "| tools offered:", flow.tools())
try:
    flow.fire("pickup_booked")          # the model tries to skip ahead
except InvalidStep as err:
    print("blocked:", err)
for event in ["order_found", "eligible", "pickup_booked"]:
    flow.fire(event)
    print("state:", flow.state, "| tools offered:", flow.tools())
print("history:", " -> ".join(flow.history))
```
