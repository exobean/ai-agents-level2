## Persist

```
import json
import tempfile
from pathlib import Path

from agentlab.persist import load, save
from shopassist.messages import Message, ToolCall

msgs = [Message("user", "I want to return order #1042"),
        Message("assistant", tool_calls=(
            ToolCall("c1", "get_order", {"order_id": "1042"}),))]
path = Path(tempfile.mkdtemp()) / "cust-7.json"
save(path, msgs, {"flow": "verified"})

data = json.loads(path.read_text())
print("keys:", list(data))
print("second message:", json.dumps(data["messages"][1])[:46], "...")
back, state = load(path)               # e.g. after a server restart
print("same messages:", back == msgs, "| state:", state)
```
