```
import json

from agentlab.mcp_client import McpClient

with McpClient() as client:          # starts the server process
    tools = client.tools()           # asks: what can you do?
    print("tools:", [t.name for t in tools])
    for direction, msg in client.log:
        print(f"{direction:<8}", json.dumps(msg)[:58], "...")
```
