## Server

```
from agentlab.mcp_server import handle

calls = [
    {"jsonrpc": "2.0", "id": 1, "method": "tools/call",
     "params": {"name": "get_order",
                "arguments": {"order_id": "1042"}}},
    {"jsonrpc": "2.0", "id": 2, "method": "tools/call",
     "params": {"name": "get_order", "arguments": {"order": "1042"}}},
    {"jsonrpc": "2.0", "id": 3, "method": "tools/delete"},
]
for msg in calls:
    reply = handle(msg)
    if "error" in reply:
        print(msg["id"], "protocol error", reply["error"]["code"])
        continue
    result = reply["result"]
    text = result["content"][0]["text"]
    print(msg["id"], "isError =", result["isError"], "|", text[:40])
```
