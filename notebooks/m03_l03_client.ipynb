## Agent

```
import textwrap

from agentlab.mcp_client import McpClient
from agentlab.rag import RagTeachingModel
from shopassist.agent import run_agent

with McpClient() as client:
    tools = client.tools()            # remote tools, local interface
    answer, _ = run_agent(RagTeachingModel(), tools, "Cite sources.",
                          "How long does a refund take?")
    calls = [m["method"] for d, m in client.log if d == "sent"]
print(textwrap.fill(answer, 68))
print("MCP methods sent:")
for method in calls:
    print("  ", method)
```
