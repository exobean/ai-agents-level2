## Chain

```
from agentlab.patterns import REPLY_CHAIN, GateFailed, chain

for message in ["Where is my order #1042?", "Where is my order?",
                "Where is #9999?"]:
    try:
        result, trace = chain(REPLY_CHAIN, {"message": message})
        print(" -> ".join(trace))
        print(f"{'':<25} {result['reply']}")
    except GateFailed as err:
        print(f"{message:<25} {err}")
```
