## Parallel

```
from agentlab.patterns import triage, vote

msg = "I am angry, where is order #1041? I want a refund"
for name, result in triage(msg).items():     # three checks at once
    print(f"{name:<7}", result if name != "order" else result["status"])

for reply in ["Your order is packed.",
              "We will definitely deliver it tomorrow."]:
    ok, votes = vote(reply)
    print("send" if ok else "hold", votes)
```
