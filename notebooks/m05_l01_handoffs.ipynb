## Handoff

```
import textwrap

from agentlab.team import TEAM, run_team

for name, agent in TEAM.items():
    print(f"{name:<8} {agent.skills}")
print()
for message in ["Where is my order #1042?",
                "Please give me my money back for #1041",
                "I am angry, my order is late"]:
    text, path = run_team(message)
    print(" -> ".join(path))
    print(textwrap.fill(text, 68, initial_indent="   ",
                        subsequent_indent="   "))
```
