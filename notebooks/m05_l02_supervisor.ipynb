## Supervise

```
import textwrap

from agentlab.team import supervise

out = supervise("Where is my order #1043? And also, is there a student "
                "discount? Can I pay cash on delivery?")
for part, worker, _ in out["parts"]:
    print(f"{worker:<8} <- {part}")
print(textwrap.fill("reply: " + out["reply"], 68))
```
