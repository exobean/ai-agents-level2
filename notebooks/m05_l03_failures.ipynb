## Failures

```
from agentlab.team import HopLimit, run_team

# 1. Ping-pong: two agents keep handing the customer to each other
try:
    run_team("I want a refund for order #1042")
except HopLimit as err:
    print("hop limit:", err)

# 2. Lost context: the order number was in an earlier message
for share in (True, False):
    ctx = {}
    run_team("Hi, my order number is #1043", ctx, share_context=share)
    text, path = run_team("Where is it?", ctx, share_context=share)
    print(f"share_context={share!s:<5}", text)
```
