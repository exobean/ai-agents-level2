## Coder

```
import shutil

from agentlab.coding import (SAMPLE, TeachingCoder, diff,
                             make_workspace, workspace_tools)
from shopassist.agent import run_agent

root = make_workspace()          # a throwaway copy of a tiny repo
answer, messages = run_agent(TeachingCoder(), workspace_tools(root),
                             "Fix the bug.", "Make the tests pass.",
                             max_steps=10)
for m in messages:
    if m.tool_calls:
        print("call:", m.tool_calls[0].name)
print("answer:", answer)
print(diff(SAMPLE, root))
shutil.rmtree(root)
```
