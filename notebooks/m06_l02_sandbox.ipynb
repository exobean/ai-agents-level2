## Sandbox

```
import shutil

from agentlab.coding import Refused, make_workspace, workspace_tools

root = make_workspace()
tools = {t.name: t for t in workspace_tools(root)}
attempts = [("read_file", {"path": "../../.ssh/id_rsa"}),
            ("run_command", {"name": "curl"}),
            ("edit_file", {"path": "shipping.py", "old": "return",
                           "new": "pass"}),
            ("read_file", {"path": "shipping.py"})]
for name, args in attempts:
    try:
        tools[name].fn(**args)
        print(f"{name:<12} allowed")
    except Refused as err:
        print(f"{name:<12} refused: {err}")
shutil.rmtree(root)
```
