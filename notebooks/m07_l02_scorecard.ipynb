## Scorecard

```
from agentlab.scorecard import GATES, failing, run_cases, scores

rows = run_cases()
score = scores(rows)
for name, value in score.items():
    op, limit = GATES[name]
    print(f"{name:<16} {value:5.2f}   gate {op} {limit}")
print("failing gates:", failing(score) or "none")
for r in rows:
    if not r["correct"]:
        print("known miss:", r["turns"][-1])
        print("   reply:", r["reply"][:56], "...")
```
